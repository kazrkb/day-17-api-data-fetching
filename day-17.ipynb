```python
# 1) Fetch products from the API (single, simple request)
response = requests.get("https://dummyjson.com/products", params={"limit": 200, "skip": 0})
print('status code ->', response.status_code)

# 2) Parse JSON into Python objects
data = response.json()
print('total products reported by API ->', data.get('total'))
products = data.get('products', [])
```

```python
# 4) Simple for-loop: build a list of rows from `products` and convert to DataFrame
rows = []
for p in products:
    row = {
        "id": p.get("id"),
        "title": p.get("title"),
        "description": p.get("description"),
        "category": p.get("category"),
        "price": p.get("price"),
        "rating": p.get("rating"),
        "stock": p.get("stock"),
    }
    rows.append(row)

loop_df = pd.DataFrame(rows)
loop_df.head()
```

```python
loop_df.to_csv("products_loop.csv", index=False)
```
